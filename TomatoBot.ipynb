```python
import importlib.util
import os

carpeta = "."

def cargar_modulo(nombre, archivo):
    ruta = os.path.join(carpeta, archivo)
    especificacion = importlib.util.spec_from_file_location(nombre, ruta)
    modulo = importlib.util.module_from_spec(especificacion)
    especificacion.loader.exec_module(modulo)
    return modulo

modulo_intenciones = cargar_modulo("Intenciones", "Intenciones.py")
modulo_procesador = cargar_modulo("ProcesadorTexto", "ProcesadorTexto.py")
modulo_identificador = cargar_modulo("IdentificarIntenciones", "IdentificarIntenciones.py")

print("¡Los módulos se cargaron correctamente!")
```

```python
procesador = modulo_procesador.ProcesadorTexto()

identificador = modulo_identificador.IdentificarIntenciones(
    modulo_intenciones.intenciones,
    procesador
)
```

```python
#Menu de Tomatobot
def mostrar_menu():
    print("*------MENU PRINCIPAL-----*\n-")
    print("*Elige una de las opciones*")
    print("1. Pedidos a Japón")
    print("2. Pedidos a Corea")
    print("3. Pagos Aduana y EMS")
    print("4. Formulario y Comprobante")
    print("5. Envío a domicilio")
    print("6. Atención personalizada")
    print("7. Restricciones")
    print("0. Salir")

def main():
    while True:
        mostrar_menu()
        leeropcion = input("Elige una opción o escribe tu pregunta: ")

        #Opción para salir del Menú
        if leeropcion == "0":
            print("¡Gracias por usar Tomatobot!")
            break

        #Verifica usuario ingresó un número
        if leeropcion.isdigit():
            opcionr = int(leeropcion)

            if opcionr == 1:
                print(
                    "Para pedir a Japón se necesita que el cliente elija "
                    "el producto en la página de Mercari Japón y enviar "
                    "el link al grupo de Mercari Japón, así se cotiza "
                    "el precio y el cliente confirma realizar el pedido."
                )

            elif opcionr == 2:
                print(
                    "Para pedir a Corea se necesita que el cliente elija "
                    "el producto en la página de Mercari Corea y enviar "
                    "el link al grupo de Mercari Corea, así se cotiza "
                    "el precio y el cliente confirma realizar el pedido."
                )

            elif opcionr == 3:
                print(
                    "Los pagos de aduana y EMS se realizan mediante la "
                    "cuenta Banco Azteca, escribiendo como concepto tu "
                    "nombre y ADUANA/EMS. Después se realiza el formulario "
                    "y se mandan las capturas en el grupo EMS y Aduana."
                )

            elif opcionr == 4:
                print(
                    "Los links de cada formulario se encuentran en la "
                    "descripción del grupo. El comprobante se tiene que "
                    "mandar al grupo de Pagos o Aduana y EMS, con tu "
                    "nombre y lo que pediste."
                )

            elif opcionr == 5:
                print(
                    "Se debe cotizar el envío a domicilio solamente "
                    "cuando se abra la fecha, la cual se estará indicando "
                    "en el grupo general."
                )

            elif opcionr == 6:
                print(
                    "Te ayudaré a contactar con una persona de atención "
                    "al cliente."
                )

            elif opcionr == 7:
                print(
                    "Estas son las restricciones que debes consultar "
                    "antes de realizar tu pedido: no nos hacemos "
                    "responsables por compras en otro grupo. Los pagos "
                    "que no se hagan a las cuentas indicadas no serán "
                    "tomados en cuenta. No se aceptan reventas bajo "
                    "ninguna circunstancia. Se vetará sin reembolso a "
                    "quien robe referencias o venda mercancía como propia."
                )

            else:
                print("La opción que ingresaste no es válida.")

        #Si no se ingresó un número de opción y se escribió texto, se ejecuta
        #la siguiente parte del código
        else:
            resultado = identificador.identificar(leeropcion)

            if resultado is not None:
                print(resultado.respuesta)
            else:
                print("No pude identificar tu pregunta. Intenta de otra manera.")
```

```python
main()
```
